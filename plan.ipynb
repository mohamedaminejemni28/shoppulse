```python
id: verifier_base_shoppulse
namespace: shoppulse.database

description: Vérifier les données réelles de ShopPulse

tasks:
  - id: compter_donnees
    type: io.kestra.plugin.jdbc.postgresql.Query
    url: "{{ envs.shoppulse_db_url }}"
    username: "{{ envs.shoppulse_db_user }}"
    password: "{{ envs.shoppulse_db_password }}"
    sql: |
      SELECT
        (SELECT count(*) FROM shop.customers) AS customers,
        (SELECT count(*) FROM shop.orders) AS orders,
        (SELECT count(*) FROM shop.events) AS events;
    fetchType: FETCH_ONE

  - id: afficher_resultat
    type: io.kestra.plugin.core.log.Log
    message: |
      Clients : {{ outputs.compter_donnees.row.customers }}
      Commandes : {{ outputs.compter_donnees.row.orders }}
      Événements : {{ outputs.compter_donnees.row.events }}
```
